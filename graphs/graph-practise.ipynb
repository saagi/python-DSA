```python
# Given:

graph = {
    0: [1, 2],
    1: [0, 3, 4],
    2: [0, 5],
    3: [1],
    4: [1],
    5: [2]
}

# Write a function:

# def dfs(start, graph):
#     ...

# that returns the set of all nodes reachable from start
```

```python
def dfs(start,graph,visited,result):
    if start in visited:
        return
    visited.add(start)
    result.append(start)
    for node in graph[start]:
        dfs(node,graph,visited,result)
    return result
```

```python
## Shortest Distance
graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "E"],
    "D": ["B", "F"],
    "E": ["C", "F"],
    "F": ["D", "E"]
}
```

```python
from collections import deque

def shortest_distance(start, graph):

    distance = {start: 0}
    queue = deque([start])

    while queue:
        node = queue.popleft()
        for neighbour in graph[node]:
            if neighbour not in distance:
                distance[neighbour] = distance[node]+1
                queue.append(neighbour)

    return distance
```
