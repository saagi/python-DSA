```python
def mergeintervals(intervals):
    intervals.sort(key=lambda x:x[0])
    result=[]
    for start,end in intervals:
        if not result or start>result[-1][1]:
            result.append([start,end])
        else:
            result[-1][1] = max(result[-1][1],end)
    return result


    
```

```python
mergeintervals([[1,2],[3,5],[4,8]])
```

```python
def insertinterval(intervals,newinterval):
    intervals.sort(key=lambda x:x[0])
    result=[]
    i=0
    newstart, newend = newinterval
    while i<len(intervals) and intervals[i][1]<newstart:
        result.append(intervals[i])
        i+=1
    while i<len(intervals) and intervals[i][0]<=newend:
        newstart = min(newstart,intervals[i][0])
        newend = max(newend,intervals[i][1])
        i+=1
    result.append([newstart,newend])
    while i<len(intervals):
        result.append(intervals[i])
        i+=1
    return result
```

```python
insertinterval([[1,2],[4,5],[6,9]],[3,8])
```

```python
def canAttendMeetings(intervals):
    intervals.sort(key=lambda x:x[0])
    result=[]
    for start,end in intervals:
        if not result or start>=result[-1][1]:
            result.append([start,end])
        else:
            return False
    return True
```

```python
canAttendMeetings([[1,2],[4,5],[6,9],[3,4]])
```

```python
# What is the minimum number of meeting rooms required so that every meeting can take place without overlapping meetings sharing the same room?

# Example 1
# intervals = [[0,30], [5,10], [15,20]]

# Visualize:

# 0       5      10      15      20              30
# |-------|-------|-------|-------|---------------|

# [0-------------------------------30]
#     [5------10]
#                     [15------20]

# At time 5:

# Meeting 1 + Meeting 2

# So we need 2 rooms.

# At time 15:

# Meeting 1 + Meeting 3

# Still only 2 rooms.

# Therefore:

# Output: 2
```

```python
def minrooms(intervals):
    intervals.sort(key=lambda x:x[0])
    if not intervals:
        return -1
    minrooms=0
    roomsending=[]
    for start,end in intervals:
        if roomsending and start>=min(roomsending):
            roomsending.remove(min(roomsending))
            roomsending.append(end)
        else:
            minrooms+=1
            roomsending.append(end)
    return minrooms
    # return minrooms
```

```python
import heapq
def minroomshead(intervals): # O(nlogn)
    if not intervals:
        return 0
    minrooms=0
    activerooms=[]
    intervals.sort(key=lambda x:x[0]) # O(nlogn)
    for start,end in intervals:#O(n) -> O(nlogn)
        if activerooms and start>=activerooms[0]:
            heapq.heappop(activerooms)#O(logn)
        else:
            minrooms+=1
        heapq.heappush(activerooms,end)#O(logn)
    return minrooms

```

```python
minroomshead([[0,5], [4,20], [5,20]])
```

```python
minroomshead([[0,5], [4,20], [5,23],[6,25]])
```
