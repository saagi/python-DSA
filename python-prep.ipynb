The mental model: BOUNDS → PROPERTY → INVARIANT → CODE → VERIFY
Recommended five-step framework that can be applied to most DSA problems, not just binary search.
5-minute pre-coding framework
1. Define the answer
   What exactly am I returning: an index, a value, a count, or an optimal value?
2. Establish valid bounds
   What's the smallest and largest possible answer? Can I prove both?
3. Identify the mathematical property
   Is the problem monotonic, sorted, overlapping, or composed of subproblems?
4. State the invariant
   What must remain true after every iteration or recursive call?
5. Define termination and verify
   When can I safely return? What happens at the boundaries?

1. What exactly must I return? — Index, value, count, boolean, or optimal value?
2. What are the valid bounds? — Prove the minimum and maximum possible answers.
3. What property enables the algorithm? — Monotonicity, sorting, overlapping intervals, repeated subproblems?
4. What is my invariant? — What remains true after every iteration?
5. Why is each discarded state safe? — Can I prove the optimal solution is not being removed?
6. When is it safe to return? — Found a valid answer, or proved the optimal answer?
7. What are the smallest counterexamples? — Single element, minimum answer, maximum answer, no solution, equal boundaries.

```python
# Problem statement
# You are given an integer array nums sorted in ascending order with distinct elements.
# Before being passed to your function, the array may have been rotated at an unknown pivot.
# For example:
# - Original: [0,1,2,4,5,6,7]
# - Rotated: [4,5,6,7,0,1,2]
# Given nums and an integer target, return the index of target if it exists, otherwise return -1.
# Constraints:
# - 1 <= nums.length <= 5000
# - -10⁴ <= nums[i], target <= 10⁴
# - All elements are unique.
# - Your algorithm must run in O(log n) time.
```

```python
from typing import List
def search(nums: List[int], target: int) -> int:
        left=0
        right=len(nums)-1
        while left<=right:
                mid = (left+right)//2
                if nums[mid]==target:
                        return mid
                if nums[left]<nums[mid]:
                        if nums[left]<=target<nums[mid]:
                                right=mid-1
                        else:
                                left=mid+1
                else:
                        if nums[mid]<target<nums[right]:
                            left=mid+1
                        else:
                            right=mid-1
        return -1
                        
                        
                
```

```python
# Question 2: Koko Eating Bananas
# LeetCode Medium
#  Binary Search on Answer

# This is an important variation of binary search for your eBay MTS-2 ML Engineer interview.
# Unlike Question 1, where we searched for an element's index, here we need to find the minimum value satisfying a constraint.
# Problem statement
# Koko has n piles of bananas. The i-th pile contains piles[i] bananas.
# Koko has h hours to finish all the bananas.
# She can choose an integer eating speed k (bananas per hour).
# Rules:
# - Each hour, she can eat bananas from only one pile.
# - If a pile contains fewer than k bananas, she finishes that pile, but cannot start another pile in the same hour.
# - She must finish all piles within h hours.
# Find the minimum integer speed k that allows her to finish within h hours.
# Examples
# Input	Hours	Output
# piles = [3,6,7,11]	8	4
# piles = [30,11,23,4,20]	5	30
# piles = [30,11,23,4,20]	6	23
```

```python
import math
def finishpiles(nums,speed):
    hours=0
    for item in nums:
        hours = hours + math.ceil(item/speed)
    return hours

def cocobanana(nums,h):
    left = 1
    right=max(nums)
    answer=right

    while left<=right:
        mid = (left+right)//2

        hours = finishpiles(nums,mid)
        print(mid,"-->",hours)

        if hours<=h:
            answer=mid
            right = mid-1

        else:
            left=mid+1

    return answer
```

```python
cocobanana([3,6,7,11],8)
```
