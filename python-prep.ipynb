The mental model: BOUNDS → PROPERTY → INVARIANT → CODE → VERIFY
Recommended five-step framework that can be applied to most DSA problems, not just binary search.
5-minute pre-coding framework
1. Define the answer
   What exactly am I returning: an index, a value, a count, or an optimal value?
2. Establish valid bounds
   What's the smallest and largest possible answer? Can I prove both?
3. Identify the mathematical property
   Is the problem monotonic, sorted, overlapping, or composed of subproblems?
4. State the invariant
   What must remain true after every iteration or recursive call?
5. Define termination and verify
   When can I safely return? What happens at the boundaries?

1. What exactly must I return? — Index, value, count, boolean, or optimal value?
2. What are the valid bounds? — Prove the minimum and maximum possible answers.
3. What property enables the algorithm? — Monotonicity, sorting, overlapping intervals, repeated subproblems?
4. What is my invariant? — What remains true after every iteration?
5. Why is each discarded state safe? — Can I prove the optimal solution is not being removed?
6. When is it safe to return? — Found a valid answer, or proved the optimal answer?
7. What are the smallest counterexamples? — Single element, minimum answer, maximum answer, no solution, equal boundaries.

```python
# Problem statement
# You are given an integer array nums sorted in ascending order with distinct elements.
# Before being passed to your function, the array may have been rotated at an unknown pivot.
# For example:
# - Original: [0,1,2,4,5,6,7]
# - Rotated: [4,5,6,7,0,1,2]
# Given nums and an integer target, return the index of target if it exists, otherwise return -1.
# Constraints:
# - 1 <= nums.length <= 5000
# - -10⁴ <= nums[i], target <= 10⁴
# - All elements are unique.
# - Your algorithm must run in O(log n) time.
```

```python
from typing import List
def search(nums: List[int], target: int) -> int:
        left=0
        right=len(nums)-1
        while left<=right:
                mid = (left+right)//2
                if nums[mid]==target:
                        return mid
                if nums[left]<nums[mid]:
                        if nums[left]<=target<nums[mid]:
                                right=mid-1
                        else:
                                left=mid+1
                else:
                        if nums[mid]<target<nums[right]:
                            left=mid+1
                        else:
                            right=mid-1
        return -1
                        
                        
                
```

```python
# Question 2: Koko Eating Bananas
# LeetCode Medium
#  Binary Search on Answer

# This is an important variation of binary search for your eBay MTS-2 ML Engineer interview.
# Unlike Question 1, where we searched for an element's index, here we need to find the minimum value satisfying a constraint.
# Problem statement
# Koko has n piles of bananas. The i-th pile contains piles[i] bananas.
# Koko has h hours to finish all the bananas.
# She can choose an integer eating speed k (bananas per hour).
# Rules:
# - Each hour, she can eat bananas from only one pile.
# - If a pile contains fewer than k bananas, she finishes that pile, but cannot start another pile in the same hour.
# - She must finish all piles within h hours.
# Find the minimum integer speed k that allows her to finish within h hours.
# Examples
# Input	Hours	Output
# piles = [3,6,7,11]	8	4
# piles = [30,11,23,4,20]	5	30
# piles = [30,11,23,4,20]	6	23
```

```python
import math
def finishpiles(nums,speed):
    hours=0
    for item in nums:
        hours = hours + math.ceil(item/speed)
    return hours

def cocobanana(nums,h):
    left = 1
    right=max(nums)
    answer=right

    while left<=right:
        mid = (left+right)//2

        hours = finishpiles(nums,mid)
        print(mid,"-->",hours)

        if hours<=h:
            answer=mid
            right = mid-1

        else:
            left=mid+1

    return answer
```

```python
cocobanana([3,6,7,11],8)
```

```python
# Question 1: Number of Islands
# LeetCode 200
# Medium
# DFS / BFS

# Problem statement
# You're given an m × n grid containing:
# - "1" — land
# - "0" — water
# An island is a group of horizontally or vertically connected land cells. Diagonal connections do not count.
# Return the number of islands.
```

```python
def numIslands(grid):

    if not grid or not grid[0]:
        return 0
    
    rows = len(grid)
    cols = len(grid[0])

    visited = set(())
    islands = 0

    directions = [
        (0,1), # right
        (0,-1), # left
        (-1,0), # up
        (1,0)  # down
    ]

    def dfs(r,c):
        if r<0 or r>=rows or c<0 or c>=cols:
            return
        if grid[r][c]=="0" or (r,c) in visited:
            return
        
        visited.add((r,c))
        for dr,dc in directions:
            nr = r+dr
            nc = c+dc
            dfs(nr,nc)

    for row in range(rows):
        for col in range(cols):
            if grid[row][col]=="1" and (row,col) not in visited:
                islands+=1
                dfs(row,col)

    return islands
```

```python
grid = [
    ["1", "1", "0"],
    ["0", "1", "0"],
    ["1", "0", "1"]
]

numIslands(grid)
```

```python
# # 1. Problem statement
# # You have a chess knight starting at coordinate (0, 0) on an infinite chessboard.
# # A knight moves in an L-shape: two squares in one direction and one square perpendicular to it.
# # Given a target coordinate (x, y), return the minimum number of knight moves required to reach it.

# Input: x = 2, y = 1
# Output: 1

# Input: x = 5, y = 5
# Output: 4

# Input: x = 0, y = 0
# Output: 0
```

```python
from collections import deque
def minknightmoves(x,y):
    x,y = abs(x),abs(y)
    directions = [
    (2,1),(2,-1),(1,2),(1,-2),
    (-2,1),(-2,-1),(-1,2),(-1,-2)
    ]
    queue = deque([(0,0,0)])
    visited = {(0,0)}
    while queue:
        presx,presy,moves = queue.popleft()
        if (presx,presy)==(x,y):
            return moves
        for dx,dy in directions:
            nx,ny,nmoves = presx+dx, presy+dy,moves+1
            if not ((-2<=nx<=x+2) and (-2<=ny<=y+2)):
                continue
            if (nx,ny) not in visited:
                visited.add((nx,ny))
                queue.append((nx,ny,nmoves))
```

```python
# #### Dynamic Programming
# Dynamic Programming is an optimization technique that solves a problem by breaking it into smaller subproblems, solving each distinct subproblem once, and reusing its result.
# The key idea is:
# Don't repeatedly solve a problem whose answer you've already calculated.

# DP is especially useful when a problem has two properties:
# 1. Overlapping subproblems: The same smaller problems appear repeatedly.
# 2. Optimal substructure: An optimal solution can be constructed from optimal solutions to smaller subproblems.
```

```python
## Fibonacci
```

```python
def traditional_fib(n):
    if n<=1:
        return n
    return traditional_fib(n-1)+traditional_fib(n-2)
```

```python
## Problem with traditional recrusive fibonacci -> O(2^n) -> Multiple calculations for a given n
# For eg: Fob(5) = Fib(4)+Fib(3) -> fib(3)+fib(2) + fib(2)+fib(1) -> for 2, calculated multiple times
```

```python
## Memoization : Top to bottom
def fib_mem(n,mem=None):
    if not mem:
        mem = {}
    if n<=1:
        return n
    if n in mem:
        return mem[n]
    mem[n] = fib_mem(n-1,mem) + fib_mem(n-2,mem)
    return mem[n]
```

```python
fib_mem(10)
```

```python
## Tabulation -> Bottom up

def fib_tab(n):
    if n<=1:
        return n
    dp = [0]*(n+1)
    dp[1] = 1
    for i in range(2,n+1):
        dp[i] = dp[i-1]+dp[i-2]
    return dp[n]
```

```python
fib_tab(10)
```

The five building blocks of a DP solution
This is the mental framework 
1. State
   What does dp[i] represent?
   Example: dp[i] is the minimum number of coins required to form amount i.

2. Transition / recurrence
   How can the current answer be constructed from smaller answers?
   Example: choose a coin c, then solve amount i-c.

3. Base case
   Which smallest subproblem has an obvious answer?
   Example: dp[0] = 0.

4. Computation order
   In what order should states be evaluated so dependencies are available?
   Example: amounts 1, 2, 3, ..., target.

5. Final answer and impossible states
   Which state should be returned? How are unreachable states represented?
   Example: return dp[amount], or -1 if it remains infinite.
These five decisions are more important than the syntax of the DP loop.

6. How to recognize DP in an interview

Certain problem descriptions suggest DP, but none of them proves DP is necessary.
Signal in question	What to investigate

Minimum or maximum cost	Can an optimal answer be built from smaller optimal answers?
Count the number of ways	Can choices be decomposed into smaller states?
Choose / skip decisions	Do recursive branches revisit the same states?
Sequences or subsequences	Can prefixes or positions define reusable states?
Unlimited choices	Can a smaller remaining target define the state?
An important distinction: not every minimum/maximum problem is DP. Koko used binary search, and Minimum Knight Moves used BFS.
The underlying structure determines the algorithm.

7. DP vs DFS vs BFS vs Greedy

Technique	Core idea	Example
DFS	Explore connected paths or states	Number of Islands
BFS	Explore states level by level	Minimum Knight Moves
Greedy	Make a locally preferred choice	Interval scheduling
DP	Reuse answers to overlapping subproblems	Coin Change
DP can even be implemented using recursion and DFS-style exploration. Memoization is what prevents repeated subproblem calculations.

```python
# Question 3 — Coin Change (LeetCode 322)
# Difficulty: Medium
# Topic: Dynamic Programming (1D DP)
# Problem Statement
# You are given an integer array coins, where each element represents a coin denomination, and an integer amount, representing a target amount.
# Return the minimum number of coins required to make up that amount.
# Rules:
# - You have an unlimited supply of each coin denomination.
# - You can use any denomination multiple times.
# - If the target amount cannot be formed using the given coins, return -1.
# - If amount = 0, return 0.
```

```python
# Input:
# coins = [1, 2, 5]
# amount = 11

# Output: 3
```

```python
def mincoins(coins,amount):
    dp=[float("inf")]*(amount+1)
    dp[0]=0
    for i in range(1,amount+1):
        for coin in coins:
            if coin<=i:
                dp[i] = min(dp[i],1+dp[i-coin])
    print(dp)
    return dp[i] if dp[i]!=float("inf") else -1
```

```python
mincoins([1,5,3,2],15)
```

```python
# Question 5 — Sliding Window Maximum
# Pattern: Sliding Window + Monotonic Deque
# Difficulty: Hard (LeetCode 239)
# Target complexity: O(n) time, O(k) space
# Problem statement
# You are given an integer array nums and an integer k.
# A sliding window of size k moves from left to right across the array, one position at a time.
# Return an array containing the maximum element in each window.
```

```python
# Input:
# nums = [1, 3, -1, -3, 5, 3, 6, 7]
# k = 3

# Output:
# [3, 3, 5, 5, 6, 7]
```

```python
from collections import deque
def maxwindow(nums,k):
    dq = deque([])
    output=[]
    for right in range(len(nums)):
        while dq and dq[0]<=right-k:
            dq.popleft()
        while dq and nums[dq[-1]]<=nums[right]:
            dq.pop()
        dq.append(right)
        if right>=k-1:
            output.append(nums[dq[0]])
    return output

```

```python
maxwindow([1, 3, -1, -3, 5, 3, 6, 7],3)
```

```python
# ## sliding window minimum

# nums = [1, 3, -1, -3, 5, 3, 6, 7]
# k = 3

# # Expected output:
# [-1, -3, -3, -3, 3, 3]
```

```python
from collections import deque
def minwindow(nums,k): #[1, 3, -1, -3, 5, 3, 6, 7], k=3
    dq = deque()
    output=[]
    for right in range(len(nums)):
        ## expired condition
        while dq and dq[0]<=right-k:
            dq.popleft()
        ## monotonically increasing
        while dq and nums[dq[-1]]>=nums[right]:
            dq.pop()
        dq.append(right)

        if right>=k-1:
            output.append(nums[dq[0]])
    return output
```

```python
# Question 7 — Word Search II (Trie + DFS)
# Difficulty: Hard — LeetCode 212
# Patterns: Trie (Prefix Tree) + DFS + Backtracking
# Problem statement
# You are given an m × n grid of characters called board and a list of strings called words.
# Find all words from the given list that can be formed on the board.
# Rules:
# - You can move horizontally or vertically: up, down, left, or right.
# - You cannot move diagonally.
# - Each cell can be used only once while forming a particular word.
# - Return all matching words without duplicates, in any order.
```

```python
class Node:
    def __init__(self):
        self.children = dict()
        self.word = None

class Trie:
    def __init__(self):
        self.root = Node()

    def insert(self,word):
        node = self.root
        for char in word:
            if char not in node.children:
                node.children[char] = Node()
            node = node.children[char]
        node.word = word

class Wordsearch:
    def search(self,board,words):

        if not board or not board[0] or words==[]:
            return []
        
        rows = len(board)
        cols = len(board[0])

        result=[]

        trie = Trie()
        for word in words:
            trie.insert(word)

        def dfs(r,c,node):
            if r<0 or r>=rows or c<0 or c>=cols or board[r][c]=='#':
                return
            char = board[r][c]

            if char not in node.children:
                return

            nextnode = node.children[char]

            if nextnode.word is not None:
                result.append(nextnode.word)
                nextnode.word = None
            original_char = board[r][c]
            board[r][c] = '#'

            dfs(r+1,c,nextnode)
            dfs(r-1,c,nextnode)
            dfs(r,c+1,nextnode)
            dfs(r,c-1,nextnode)

            board[r][c] = original_char

        for r in range(rows):
            for c in range(cols):
                dfs(r,c,trie.root)

        return result

    


```
