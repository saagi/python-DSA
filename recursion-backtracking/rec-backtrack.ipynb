A recursive function has two execution phases:

Going down: Recursive calls are added to the call stack.

Coming back: Execution resumes after each recursive call returns.

This is particularly important when we start backtracking.

```python
def printnum(n):
    if n==0:
        return n
    printnum(n-1)
    print(n)
```

```python
# 5 -> 4 -> 3 -> 2 -> 1 -> 0 ->
```

```python
printnum(5)
```

```python
def solve(n):
    if n == 0:
        return

    print(n)

    solve(n - 1)

    print(n)

solve(3)
```

```python
def subsets(nums):
    result = []
    def backtrack(index,path): #0,[]
        # if index==len(nums):
        result.append(path.copy())
            

        for i in range(index, len(nums)): # 0,1
            path.append(nums[i]) # [1]
            backtrack(i,path) #(1,[1])

            path.pop()

    backtrack(0,[])
    return result
```
