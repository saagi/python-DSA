```python
# 4. Minimum Size Subarray Sum

# Classic interview problem.

# Given:

# nums = [2,3,1,2,4,3]
# target = 7

# Find the minimum-length subarray whose sum ≥ target.

# Answer:

# 2
```

```python
def minsizesubarray(nums,target):
    if sum(nums)<target:
        return -1
    left = 0
    windowsum = 0
    minlen = float("inf")

    for right in range(len(nums)):
        windowsum+=nums[right]
        while windowsum>=target:
            minlen = min(minlen,right-left+1)
            windowsum-=nums[left]
            left+=1
    return (minlen if minlen!=float("inf") else 0)

```

```python
minsizesubarray([2,3,1,2,4,3],7)
```

```python
# Question 3 — Maximum Number of Vowels

# Given a string s and an integer k, find the maximum number of vowels in any substring of length k.

# Vowels are:

# a, e, i, o, u
# Example
# s = "a
# bciiidef"
# k = 3

# Output:

# 3
```

```python
def maxvowels(s,k):
    left=0
    maxv=float("-inf")
    vowels=['a','e','i','o','u']
    window_vowels=0
    for right in range(len(s)):
        if s[right] in vowels:
            window_vowels+=1
        if right-left+1 == k:
            maxv = max(maxv,window_vowels)
            if s[left] in vowels:
                window_vowels-=1
            left+=1
    return (maxv if maxv!=float("-inf") else 0)

maxvowels("abciiidef",3)
```
